**READ DATASET**

```python
df = spark.read.csv(
    "/content/drive/MyDrive/Pyspark_Project/data/data.csv",
    header=True,
    inferSchema=True
)
```

**COUNTING NULL VALUES**

```python
from pyspark.sql.functions import col

for column in df.columns:
    print(column, ":", df.filter(col(column).isNull()).count())
```

**COUNTING DUPLICATE ROWS**

```python
print("Total Rows :", df.count())
print("Distinct Rows :", df.distinct().count())
```

**COUNT UNIQUE CUSTOMER ID(S)**

```python
df.select("CustomerID").distinct().count()
```

**COUNT UNIQUE COUNTRIES**

```python
df.select("Country").distinct().count()
```

**DATE RANGE**

```python
print(df.select(
    min("InvoiceDate").alias("Start Date"),
    max("InvoiceDate").alias("End Date")
).show())
```
